```python
import io
import csv
import numpy as np

# Messy raw dataset
raw_csv_data = """product_id,price,quantity,category
P001, $120.50 ,10,Electronics
P002, 45.00, ,Clothing
P003, $350.00,5,electronics
P004, missing,12,Home
P005, 15.75,3, clothing
"""
print("Environment ready and raw data loaded!")
```

```python
def clean_price(price_str):
    cleaned = price_str.replace('$', '').strip()
    if cleaned.lower() in ['missing', '', 'na', 'null']:
        return 0.0
    try:
        return float(cleaned)
    except ValueError:
        return 0.0

def clean_quantity(qty_str):
    cleaned = qty_str.strip()
    if cleaned == '' or not cleaned.isdigit():
        return 0
    return int(cleaned)

def clean_category(category_str):
    return category_str.strip().capitalize()

print("Cleaning functions compiled successfully!")
```

```python
data_stream = io.StringIO(raw_csv_data.strip())
reader = csv.DictReader(data_stream)

cleaned_dataset = []
for row in reader:
    cleaned_dataset.append({
        'product_id': row['product_id'].strip(),
        'price': clean_price(row['price']),
        'quantity': clean_quantity(row['quantity']),
        'category': clean_category(row['category'])
    })

print("--- Cleaned Structured Data ---")
for row in cleaned_dataset:
    print(row)
```

```python
# Convert lists to high-performance NumPy arrays
prices = np.array([item['price'] for item in cleaned_dataset])
quantities = np.array([item['quantity'] for item in cleaned_dataset])

# Vectorized math: multiplies matching elements instantly
total_revenues = prices * quantities

print("--- NumPy Operational Metrics ---")
print(f"Total Revenue: ${np.sum(total_revenues):.2f}")
print(f"Average Price: ${np.mean(prices[prices > 0]):.2f}")
print(f"Total Inventory Count: {np.sum(quantities)}")
```
